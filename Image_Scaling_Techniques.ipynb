```python
from google.colab import files

uploaded = files.upload()
```

```python
import cv2
import os
import matplotlib.pyplot as plt
```

```python
filename = list(uploaded.keys())[0]

print("Uploaded image:", filename)
```

```python
image = cv2.imread(filename)

height, width = image.shape[:2]

print("Original Width:", width)
print("Original Height:", height)
```

```python
bigger_image = cv2.resize(
    image,
    (1000, 800),
    interpolation=cv2.INTER_LINEAR
)

cv2.imwrite('bigger_image.jpg', bigger_image)

print("Bigger image size:", bigger_image.shape)
print("Bigger image created")
```

```python
smaller_image = cv2.resize(
    image,
    (250, 200),
    interpolation=cv2.INTER_AREA
)

cv2.imwrite('smaller_image.jpg', smaller_image)

print("Smaller image size:", smaller_image.shape)
print("Smaller image created")
```

```python
plt.figure(figsize=(5, 4))
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")
plt.show()

plt.figure(figsize=(10, 8))
plt.imshow(cv2.cvtColor(bigger_image, cv2.COLOR_BGR2RGB))
plt.title("Bigger Image")
plt.axis("off")
plt.show()

plt.figure(figsize=(3, 2))
plt.imshow(cv2.cvtColor(smaller_image, cv2.COLOR_BGR2RGB))
plt.title("Smaller Image")
plt.axis("off")
plt.show()
```
